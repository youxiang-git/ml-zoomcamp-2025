Question 7. Sum of weights (1 point)

```python
cars_from_asia = df[df["origin"] == "Asia"]
first_7_cars_from_asia = cars_from_asia[["vehicle_weight", "model_year"]][:7]
print(first_7_cars_from_asia)
```

```python
X = first_7_cars_from_asia.to_numpy()
print(X.shape)
print(X)
```

```python
XT = X.T
print(XT.shape)
print(XT)
```

```python
XTX = np.matmul(XT, X)
print(XTX.shape)
print(XTX)
```

```python
inverse_XTX = np.linalg.inv(XTX)
print(inverse_XTX)
```

```python
sanity_check_identity = np.matmul(inverse_XTX, XTX)
print(sanity_check_identity)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
print(y)
```

```python
w = np.matmul(np.matmul(inverse_XTX, XT), y)
print(w)
```

```python
print(w.sum())
```
